***Section 2 — Data Cleaning and Preprocessing***

```python
import pandas as pd
#Handling Missing Values 
```

```python
print("Missing values percentage:\n", df.isnull().sum() / len(df) * 100)
```

```python
age_median = int(df['age'].dropna().astype('int64').median())
absences_median = int(df['absences'].dropna().astype('int64').median())
g3_median = int(df['G3'].dropna().astype('int64').median())

df.fillna({'age': age_median, 'absences': absences_median, 'G3': g3_median}, inplace=True)
```

```python
sex_mode = df['sex'].mode()[0]
address_mode = df['address'].mode()[0]
df.fillna({'sex': sex_mode, 'address': address_mode}, inplace=True)

print("\nMissing values after treatment:\n", df.isnull().sum())
```

```python
#Fixing Inconsistent Formats
```

```python
print("Unique 'sex' values before cleaning:", df['sex'].unique())
```

```python
df['sex'] = df['sex'].str.lower().str.strip()
df['address'] = df['address'].str.lower().str.strip()
```

```python
df.replace({'sex': {'f': 'female', 'm': 'male'}}, inplace=True)
df.replace({'address': {'u': 'urban', 'r': 'rural'}}, inplace=True)
```

```python
print("Unique 'sex' values after cleaning:", df['sex'].unique())
print("Formats standardized.")
```

```python
print("***Handling Duplicates***")
print("Duplicates found:", df[df.duplicated()].shape[0])
```

```python
df.drop_duplicates(keep='first', inplace=True)
print("Dataset shape after duplicate removal:", df.shape)
```

```python
print("***Correcting datatype***")
df['age'] = df['age'].astype('int64')
df['absences'] = df['absences'].astype('int64')
df['G3'] = df['G3'].astype('int64')
print(df[['age', 'absences', 'G3']].dtypes)
```

```python
import numpy as np
from scipy import stats

print("***Handling Outliers***")
z_scores = np.abs(stats.zscore(df[['age', 'absences', 'G3']]))

df_clean = df[~(z_scores > 3).any(axis=1)]
print("Cleaned dataset shape:", df_clean.shape)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

sns.set_style('whitegrid')

df = pd.read_csv('cleaned_student_data.csv')

print(f"Dataset Shape: {df.shape[0]} rows & {df.shape[1]} columns")
df.head()
```

```python
# ==========================================
# STEP 2: SUMMARY STATISTICS & INSPECTION
# ==========================================
print("--- Data Types & Non-Null Values ---")
df.info()

print("\n--- Summary Statistics (Numerical Features) ---")
display(df.describe())
```

```python
# ==========================================
# STEP 3: UNIVARIATE ANALYSIS & OUTLIERS
# ==========================================
# A. Target Variable (G3) & Grades Distribution
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
sns.histplot(df['G1'], kde=True, ax=axes[0], color='skyblue').set_title('G1 Grade Distribution')
sns.histplot(df['G2'], kde=True, ax=axes[1], color='salmon').set_title('G2 Grade Distribution')
sns.histplot(df['G3'], kde=True, ax=axes[2], color='purple').set_title('G3 Final Grade Distribution')
plt.tight_layout()
plt.show()
```

```python
# B. IQR-based Outlier Detection
num_cols = ['age', 'absences', 'G1', 'G2', 'G3']
print("--- IQR OUTLIER DETECTION ---")
for col in num_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    outliers = df[(df[col] < (Q1 - 1.5 * IQR)) | (df[col] > (Q3 + 1.5 * IQR))]
    print(f"Column '{col}': {len(outliers)} outliers detected")

# Outliers Visualisation (Boxplots)
plt.figure(figsize=(10, 4))
sns.boxplot(data=df[num_cols], palette='Set3')
plt.title('Outliers Visualisation Across Numerical Features')
plt.show()
```

```python
# C. Categorical Feature Distributions
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
sns.countplot(x='sex', data=df, ax=axes[0], palette='pastel').set_title('Gender Distribution')
sns.countplot(x='address', data=df, ax=axes[1], palette='pastel').set_title('Address (Urban vs Rural)')
sns.countplot(x='studytime', data=df, ax=axes[2], palette='pastel').set_title('Weekly Study Time')
plt.tight_layout()
plt.show()
```

```python
# ==========================================
# STEP 3: BIVARIATE ANALYSIS 
# ==========================================
# A. Past Grades (G1, G2) vs Final Grade (G3)
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
sns.regplot(x='G1', y='G3', data=df, ax=axes[0], color='blue', scatter_kws={'alpha':0.5})
axes[0].set_title('G1 vs Final Grade (G3)')

sns.regplot(x='G2', y='G3', data=df, ax=axes[1], color='green', scatter_kws={'alpha':0.5})
axes[1].set_title('G2 vs Final Grade (G3)')
plt.tight_layout()
plt.show()
```

```python
# B. Impact of Key Categorical Features on Final Grade (G3)
fig, axes = plt.subplots(1, 3, figsize=(15, 5))
sns.boxplot(x='studytime', y='G3', data=df, ax=axes[0], palette='Blues').set_title('Study Time vs G3')
sns.boxplot(x='failures', y='G3', data=df, ax=axes[1], palette='Reds').set_title('Past Failures vs G3')
sns.boxplot(x='higher', y='G3', data=df, ax=axes[2], palette='Greens').set_title('Wants Higher Ed vs G3')
plt.tight_layout()
plt.show()
```

```python
# ==========================================
# STEP 5: MULTIVARIATE ANALYSIS
# ==========================================

# A. Correlation Heatmap
plt.figure(figsize=(10, 7))
numeric_df = df.select_dtypes(include=['int64', 'float64'])
sns.heatmap(numeric_df.corr(), annot=True, fmt='.2f', cmap='coolwarm', linewidths=0.5)
plt.title('Multivariate Correlation Heatmap')
plt.show()

```

```python
# B. Study Time vs G3 Segmented by Gender
plt.figure(figsize=(8, 5))
sns.boxplot(x='studytime', y='G3', hue='sex', data=df, palette='Set2')
plt.title('Study Time vs G3 (Grouped by Gender)')
plt.show()
```

**Section 4 — Visualizations**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style('whitegrid')
df = pd.read_csv('cleaned_student_data.csv')
print("Dataset loaded for visualization:", df.shape)
```

```python
import os
os.makedirs('reports', exist_ok=True)

#Line Chart
plt.figure(figsize=(8, 5))
study_trend = df.groupby('studytime')['G3'].mean().reset_index()
sns.lineplot(x='studytime', y='G3', data=study_trend, marker='o', color='b', linewidth=2.5)

plt.title('Trend of Average Final Grade (G3) by Study Time')
plt.xlabel('Weekly Study Time Category')
plt.ylabel('Average Final Grade (G3)')

plt.savefig('reports/linechart_studytime_g3.png', dpi=150, bbox_inches='tight')
plt.show()
```

This line chart illustrates a positive trend between weekly study time and final academic performance. Students who dedicate more hours to studying consistently achieve higher average final grades (G3), indicating that increased study time yields positive academic returns.

```python
# Chart 2: Bar Chart (Categorical Comparison)
plt.figure(figsize=(8, 5))
sns.barplot(x='sex', y='G3', data=df, hue='sex', palette='viridis', legend=False)
plt.title('Average Final Grade (G3) by Gender')
plt.xlabel('Gender (sex)')
plt.ylabel('Average Final Grade (G3)')

plt.savefig('reports/barplot_sex_g3.png', dpi=150, bbox_inches='tight')
plt.show()
```

This bar chart compares the average final grades across gender demographics. It allows us to observe baseline performance differences between male and female students to ensure our predictive insights account for demographic variances.

```python
# Chart 3: Correlation Heatmap (Numerical Relationships)
plt.figure(figsize=(10, 8))
numeric_df = df.select_dtypes(include=['float64', 'int64'])
sns.heatmap(numeric_df.corr(), annot=False, cmap='coolwarm', linewidths=0.5)

plt.title('Correlation Heatmap of Numerical Features')

plt.savefig('reports/correlation_heatmap.png', dpi=150, bbox_inches='tight')
plt.show()
```

The heatmap reveals that previous grades (G1 and G2) are the strongest positive predictors of the final grade (G3), with correlations exceeding 0.80. Additionally, it highlights that past class failures have a notable negative correlation, serving as a strong warning sign for poor final outcomes.

```python
# Chart 4: Scatter Plot (G1 vs G3 Grades)
plt.figure(figsize=(8, 5))
sns.scatterplot(x='G1', y='G3', data=df, hue='sex', palette='Set1', alpha=0.8)

plt.title('First Period Grade (G1) vs Final Grade (G3)')
plt.xlabel('G1 Score')
plt.ylabel('G3 Score')

plt.savefig('reports/scatter_g1_g3.png', dpi=150, bbox_inches='tight')
plt.show()
```

This scatter plot demonstrates a clear, strong linear relationship between a student's first-period grade (G1) and their final grade (G3). It confirms that early academic success is a critical and reliable indicator of how well a student will finish the year.

```python
# Chart 5: Boxplot (Distribution & Outliers)
plt.figure(figsize=(8, 5))
sns.boxplot(x='higher', y='G3', data=df, hue='higher', palette='Set2', legend=False)

plt.title('Final Grades Distribution by Higher Education Aspiration')
plt.xlabel('Wants Higher Education (higher)')
plt.ylabel('Final Grade (G3)')

plt.savefig('reports/boxplot_higher_g3.png', dpi=150, bbox_inches='tight')
plt.show()
```

The boxplot clearly shows that students who wish to pursue higher education ("yes") maintain a significantly higher median final grade than those who do not ("no"). Furthermore, the "yes" group displays a higher overall grade range, suggesting that academic ambition directly translates to better performance.

Based on the exploratory data analysis and visualizations generated from the cleaned student dataset, our team identified the following five key insights:

1. **Early Success is the Strongest Predictor:** The scatter plot and correlation heatmap confirm a massive positive correlation (>0.80) between early grades (G1, G2) and final grades (G3). This indicates that early intervention during the first grading period is critical for at-risk students.
2. **Study Time Yields Positive Returns:** The line chart demonstrates a clear upward trend; students who dedicate more weekly hours to studying consistently achieve higher average final grades.
3. **Past Failures are a Major Warning Sign:** The heatmap highlights that past class failures have a strong negative correlation with final mathematics grades, serving as the biggest red flag for ongoing academic struggles.
4. **Academic Ambition Drives Success:** The boxplot analysis reveals that students who intend to pursue higher education maintain a significantly higher median final grade and overall performance range than those who do not.
5. **Demographic Baselines:** The bar chart analysis allowed us to observe baseline performance differences across gender, which ensures that future predictive models and targeted interventions can account for demographic variances.
